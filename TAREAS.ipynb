**ENCONTRAR LA MEDIA ARMONICA**

```python
# ==========================================
# Programa para calcular la Media Armónica
# ==========================================

print("--- Calculadora de Media Armónica ---")

cantidad_datos = int(input("¿Cuántos números vas a ingresar?: "))

numeros = []

for i in range(cantidad_datos):
    valor = float(input(f"Ingresa el número {i + 1}: "))

    while valor == 0:
        print("¡Error! La media armónica no permite ceros (no se puede dividir entre 0).")
        valor = float(input(f"Ingresa nuevamente el número {i + 1} (diferente de 0): "))

    numeros.append(valor)

suma_reciprocos = 0.0

for num in numeros:
    suma_reciprocos += 1 / num

media_armonica = cantidad_datos / suma_reciprocos

print("\n------------------------------------")
print(f"Datos ingresados: {numeros}")
print(f"La Media Armónica es: {media_armonica:.4f}")
print("------------------------------------")
```

***PROGRAMA PARA VERIFICAR PALINDROMOS***

```python
# ========================================================
# Programa para verificar Palíndromos (Uso de POO)
# ========================================================

class ValidadorPalindromo:
    def __init__(self, texto):
        self.texto_original = texto
        self.texto_limpio = self._limpiar_texto(texto)

    def _limpiar_texto(self, texto):
        """Función interna para normalizar el texto introducido."""
        texto_procesado = texto.lower().replace(" ", "")
        return texto_procesado

    def es_palindromo(self):
        """Determina si la palabra/frase se lee igual al derecho y al revés."""
        texto_invertido = self.texto_limpio[::-1]
        return self.texto_limpio == texto_invertido

    def mostrar_resultado(self):
        """Imprime la respuesta formateada en consola."""
        print("\n------------------------------------")
        print(f"Texto ingresado: '{self.texto_original}'")

        if self.es_palindromo():
            print("Resultado: ¡SÍ es un palíndromo!")
        else:
            print("Resultado: NO es un palíndromo.")

        print("------------------------------------")



print("--- Verificador de Palíndromos ---")


entrada_usuario = input("Ingresa una palabra o frase: ")

verificador = ValidadorPalindromo(entrada_usuario)

verificador.mostrar_resultado()
```
