```python
# 1. Definimos la función del factorial
def calcular_factorial(numero):
    n = int(numero)  # El factorial solo funciona con enteros

    # Caso especial: el factorial de 0 y 1 siempre es 1
    if n == 0 or n == 1:
        return 1

    # Variable para ir acumulando el resultado de las multiplicaciones
    resultado = 1

    # Bucle que va desde 1 hasta 'n'
    for i in range(1, n + 1):
        resultado = resultado * i  # Multiplica acumulativamente

    return resultado

# 2. Pedimos el número al usuario
num = input("Ingrese un número entero para calcular su factorial: ")

# 3. Mostramos el resultado
print("El factorial de", num, "es:", calcular_factorial(num))
```

Calcular el suelo de un empleado. horas * valor de horas * 30 días, las deducciones:ISSS, AFP, ISR.

```python
# 1. Definimos la función principal de cálculo
def calcular_planilla_sv(horas_diarias, valor_hora):
    # Convertimos los datos de entrada a números decimales
    h = float(horas_diarias)
    v = float(valor_hora)

    # --- SUELDO BRUTO ---
    sueldo_bruto = h * v * 30

    # --- DEDUCCIÓN ISSS (3% con techo de $30.00) ---
    isss = sueldo_bruto * 0.03
    if isss > 30.00:
        isss = 30.00

    # --- DEDUCCIÓN AFP (7.25%) ---
    afp = sueldo_bruto * 0.0725

    # --- BASE PARA EL IMPUESTO SOBRE LA RENTA (ISR) ---
    sueldo_imponible = sueldo_bruto - isss - afp

    # --- CÁLCULO DE ISR POR TRAMOS (Mensual) ---
    if sueldo_imponible <= 472.00:
        isr = 0.0
    elif sueldo_imponible <= 895.24:
        isr = ((sueldo_imponible - 472.00) * 0.10) + 17.67
    elif sueldo_imponible <= 2038.10:
        isr = ((sueldo_imponible - 895.24) * 0.20) + 60.00
    else: # Más de $2038.10
        isr = ((sueldo_imponible - 2038.10) * 0.30) + 288.57

    # --- SUELDO NETO (Lo que recibe en mano) ---
    total_deducciones = isss + afp + isr
    sueldo_neto = sueldo_bruto - total_deducciones

    #resultados detallados
    print("\n--- RESUMEN DE PLANILLA ---")
    print(f"Sueldo Bruto: ${sueldo_bruto:.2f}")
    print(f"(-) ISSS (3%): ${isss:.2f}")
    print(f"(-) AFP (7.25%): ${afp:.2f}")
    print(f"(-) ISR (Renta): ${isr:.2f}")
    print("----------------------------")
    print(f"Total Descuentos: ${total_deducciones:.2f}")
    print(f"SUELDO NETO A RECIBIR: ${sueldo_neto:.2f}")


#Pedimos los datos al usuario
horas = input("Ingrese la cantidad de horas trabajadas al día: ")
precio_hora = input("Ingrese el valor monetario de la hora ($): ")

# 3. Llamamos a la función
calcular_planilla_sv(horas, precio_hora)
```

```python
def conversor_volumen():
    conversiones = {
        "1": ("Litros", "Mililitros", 1000.0),
        "2": ("Mililitros", "Litros", 0.001),
        "3": ("Galones (US)", "Litros", 3.78541),
        "4": ("Litros", "Galones (US)", 0.264172),
        "5": ("Metros cúbicos", "Litros", 1000.0),
        "6": ("Litros", "Metros cúbicos", 0.001),
        "7": ("Onzas líquidas", "Mililitros", 29.5735),
        "8": ("Mililitros", "Onzas líquidas", 0.033814),
        "9": ("Tazas", "Litros", 0.236588),
        "10": ("Litros", "Tazas", 4.22675)
    }


    while True:
        print("\n" + "="*35)
        print("     CONVERSOR DE VOLUMEN     ")
        print("="*35)


        for numero, datos in conversiones.items():
            unidad_origen = datos[0]
            unidad_destino = datos[1]
            print(f"{numero}. De {unidad_origen} a {unidad_destino}")

        print("-" * 35)


        opcion = input("Elige el número de la conversión que deseas realizar (1-10): ")


        if opcion in conversiones:

            try:
                unidad_origen = conversiones[opcion][0]
                unidad_destino = conversiones[opcion][1]
                factor = conversiones[opcion][2]

                cantidad = float(input(f"\nIngresa la cantidad de {unidad_origen} a convertir: "))


                resultado = cantidad * factor


                print(f"\n RESULTADO: {cantidad} {unidad_origen} equivalen a {resultado:.4f} {unidad_destino}")

            except ValueError:
                print("\n Error: Por favor, ingresa un número válido (ej. 10 o 5.5).")
        else:
            print("\n Opción no válida. Debes elegir un número del 1 al 10.")


        continuar = input("\n¿Deseas calcular otro volumen? (s/n): ").lower()


        if continuar != 's':
            print("\n¡Gracias por usar el conversor! Hasta la próxima. ")
            break

conversor_volumen()
```

```python

from functools import reduce
class estadistica:
  def media(self, serie):
    suma = reduce(lambda acum, num: acum+ float(num), serie, 0)
    return suma/len(serie)

  def varianza(self, serie):
    media = self.media(serie)
    #4 + 0 + 4 + 0 + 0 + 16= 24/6 = 4.0
    suma = reduce(lambda acum, num: acum + (float(num)-media)**2, serie, 0)
    return suma/len(serie)

  def desviacion(self, serie):
    return self.varianza(serie)**0.5
```

```python
#Hacer un programa en python que lea una serie de numeros introducida por el usuario
#y que calcule la media aritmetica
objEstadistica = estadistica()
serie = input("Serie: ").split(",")
print("Media: ", objEstadistica.media(serie))
print("Varianza: ", objEstadistica.varianza(serie))
print("Desviacion: ", objEstadistica.desviacion(serie))
```
